## 8. Why this metric matches the real decision

> **Precision@K** directly connects to the business action: content reviewers can
only review a limited number of pages per cycle. If we review the top K prioritized
pages, Precision@K tells us what fraction of those K pages are actually declining
(and thus genuine refresh candidates). 

>A high Precision@K means the model is putting the most promising pages at the
top of the list, so the reviewer spends their limited time on pages most likely
to benefit from a refresh. This matters more than overall accuracy across all
30,000 pages, because the reviewer cannot act on all pages.

>K is described generically as the number of pages the content team can
realistically review in a given cycle, because the exact capacity threshold is a
policy choice that the business makes, not a universal constant. The metric
design acknowledges this by using a relative (top-K) rather than absolute measure.
